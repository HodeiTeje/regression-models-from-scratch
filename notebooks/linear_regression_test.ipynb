#### In this notebook we will test the implementation of the linear regression model to check whether it works correctly or not. For this purpose, we will generate dummy data that follows $y = 3x + 5$, but we will add some noise. 

```python
import numpy as np
import matplotlib.pyplot as plt

from src.linear_regression import LinearRegression
```

```python
np.random.seed(42)

X = np.linspace(0, 10, 100).reshape(-1, 1)
y = 3 * X.ravel() + 5 + np.random.normal(0, 2, 100)

print("Shape of X:", X.shape)
print("Shape of y:", y.shape)

print("First 5 values of X:")
print(X[:5])

print("First 5 values of y:")
print(y[:5])
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], y, label="Generated data")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Synthetic dataset")
plt.legend()
plt.grid(True)
plt.show()
```

```python
lr_model = LinearRegression()

costs = lr_model.fit(X, y)

print("Learned weights: ", lr_model.weights)
print("Learned bias: ", lr_model.bias)

print("Final cost: ", lr_model.compute_cost(X, y))
```

```python

plt.figure(figsize=(8, 5))
plt.plot(costs)
plt.xlabel("Iteration")
plt.ylabel("Mean Squared Error")
plt.title("Training cost")
plt.grid(True)
plt.show()
```

```python
predictions = lr_model.predict(X)

plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], y, label="Generated data")
plt.plot(X[:, 0], predictions, color="red", label="Learned regression")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear regression result")
plt.legend()
plt.grid(True)
plt.show()
```
